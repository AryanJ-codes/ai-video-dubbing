## Cell 1 — Install (then manually restart via Runtime menu)

```python
import os

# System dependencies
os.system('apt-get install -y ffmpeg -q')

# Always pull latest code (removes stale clones from old sessions)
os.system('rm -rf ai-video-dubbing')
os.system('git clone https://github.com/AryanJ-codes/ai-video-dubbing.git')

# Install Python requirements
# Note: torch/torchaudio/torchvision are pre-installed in Colab with CUDA
ret = os.system('pip install -r ai-video-dubbing/requirements.txt -q')

if ret == 0:
    print()
    print('=' * 50)
    print('✅ Done! Now go to Runtime → Restart session')
    print('   Then run Cells 2 → 3 → 4')
    print('=' * 50)
else:
    print('❌ Installation had errors. Check output above.')
```

## Cell 2 — Setup (run after restarting)

```python
import sys, os

PROJECT_DIR = '/content/ai-video-dubbing'
sys.path.insert(0, PROJECT_DIR)
os.chdir(PROJECT_DIR)

# Pre-accept Coqui TTS license (prevents interactive prompt)
os.environ['COQUI_TOS_AGREED'] = '1'

# Verify packages are importable
try:
    import yt_dlp, whisper, TTS, deep_translator
    print('✅ All packages found. Ready to run!')
except ImportError as e:
    print(f'⚠️  Missing: {e}. Re-installing...')
    os.system('pip install -r requirements.txt -q')
    print('✅ Done. Please Re-run this cell once.')
```

## Cell 3 — Dub a video

```python
YOUTUBE_URL = "https://www.youtube.com/watch?v=M7FIvfx5J10"  # ← Change this

from main import run_pipeline

def show_progress(progress, stage, message, eta_seconds):
    bar = int(progress / 5)
    eta = f" | ETA: {int(eta_seconds)}s" if eta_seconds and eta_seconds > 0 else ""
    print(f"[{'█' * bar}{'░' * (20 - bar)}] {int(progress):3d}%  [{stage}] {message}{eta}", end='\r')

print(f"🎬 Starting: {YOUTUBE_URL}\n")
output_path = run_pipeline(YOUTUBE_URL, status_updater=show_progress)
print(f"\n\n✅ Done! Saved to: {output_path}")
```

## Cell 4 — Play & Download

```python
from IPython.display import Video, display
from google.colab import files

display(Video(output_path, embed=True, width=720))
files.download(output_path)
```
