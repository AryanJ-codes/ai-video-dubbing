## Cell 3 — Dub a video

Change `YOUTUBE_URL` below and run.

```python
YOUTUBE_URL = "https://www.youtube.com/watch?v=M7FIvfx5J10"  # ← Change this

from main import run_pipeline

def show_progress(progress, stage, message, eta_seconds):
    bar = int(progress / 5)
    eta = f" | ETA: {int(eta_seconds)}s" if eta_seconds and eta_seconds > 0 else ""
    print(f"[{'█' * bar}{'░' * (20 - bar)}] {int(progress):3d}%  [{stage}] {message}{eta}", end='\r')

print(f"🎬 Starting: {YOUTUBE_URL}\n")
output_path = run_pipeline(YOUTUBE_URL, status_updater=show_progress)
print(f"\n\n✅ Done! Saved to: {output_path}")
```

## Cell 4 — Play & Download

```python
from IPython.display import Video, display
from google.colab import files

display(Video(output_path, embed=True, width=720))
files.download(output_path)
```
